## Equilbrage-Imbalanced Learning

```python
#Data Transormation
# Ensure that all columns are NUMERICAL to be able to use imblearn
from sklearn.preprocessing import LabelEncoder

le=LabelEncoder()
for col in df.columns:
    df[col]=le.fit_transform(df[col])  # Apply the LabelEncoder()
    print(col, "column:", df[col].value_counts().to_dict())
```

```python
#df['banking_crisis'] colonne prédite
df['banking_crisis'].value_counts().to_dict() 
#equilibrage is only concerned with binary columns where there is an imbalance between number of zeros and number of ones
```

```python
from imblearn.over_sampling import SMOTE #no need to make research about the details of SMOTE, understanding the concept only is enough 

#splitting colonne predite (y) from other cols (x)
X=df.drop(columns=['banking_crisis'])
y=df['banking_crisis'] #colonne prédite/ target

#initialize & apply the SMOTE algorithm
smote = SMOTE(random_state=42)
X_resampled_smote, y_resampled_smote = smote.fit_resample(X, y) #we will put the results of fit resample in two variables that are balanced

#concatenate & create the new balanced DataFrame
df_resampled = pd.concat([X_resampled_smote, y_resampled_smote], axis=1)   # axis=1  ==> column level: concat them au niveau des colonnes
df = df_resampled

#this code always stays the same (change name of columns only)
```

## Outliers-Valeurs abérantes

```python
#to replace outliers by median, applicable to all datasets
from scipy.stats import zscore
import numpy as np

numeric_columns = df.select_dtypes(include=[np.number])
z_scores = np.abs(zscore(numeric_columns))
cleaned_dataset = df.copy()

#iterate over each numeric column
for col in numeric_columns.columns:
    median = numeric_columns[col].median()
    outliers = z_scores[col] > 3   #
    cleaned_dataset.loc[outliers, col] = median

df = cleaned_dataset
```
